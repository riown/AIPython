```python
# 사칙연산(+, -, *, /)을 하는 아주 간단한 프로그램
# (변수, 연산자, 출력문만 사용)

# 1) 두 수를 변수에 저장
a = 12
b = 3

# 2) 연산 결과를 변수에 저장
add_result = a + b   # 덧셈
sub_result = a - b   # 뺄셈
mul_result = a * b   # 곱셈
div_result = a / b   # 나눗셈 (파이썬에서는 / 결과가 보통 실수)

# 3) 출력문으로 결과 출력
print("a =", a)
print("b =", b)
print("a + b =", add_result)
print("a - b =", sub_result)
print("a * b =", mul_result)
print("a / b =", div_result)
```
