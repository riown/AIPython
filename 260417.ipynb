```python
cafe_menu = {
    "아메리카노": 4500,
    "카페라떼": 5200,
    "카푸치노": 5200,
    "바닐라라떼": 5900,
    "카라멜마키아토": 6100,
    "카페모카": 6300,
    "에스프레소": 4000,
    "콜드브루": 5500,
    "자몽에이드": 6000,
    "초코라떼": 5800
}

order = {}

while True:
    # 1) 첫 주문 포함, 매번 메뉴 전체 출력
    print("\n=== 카페 메뉴(전체) ===")
    for name, price in cafe_menu.items():
        print(f"{name} : {price}원")
    print("=======================")

    # 2) 그 다음에 입력 받기
    menu = input("메뉴를 입력하세요: ")
    count = int(input("개수를 입력하세요: "))


    # 4) 계속/종료
    again = input("또 주문할까요? (계속/종료): ")
    if again == "종료":
        break

# 5) 최종 출력: 각 메뉴의 총 개수 + 총 금액
print("\n=== 최종 주문 내역 ===")
total_price = 0
for menu_name, qty in order.items():
    print(f"{menu_name} : 총 {qty}개")
    total_price += cafe_menu[menu_name] * qty

print(f"총 금액: {total_price}원")
```
