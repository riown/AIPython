```python
fruits = ["strawberry", "lemon", "apple"]
```

```python
fruits
```

```python
len(fruits)
```

```python
for fruit in fruits :
  print(fruit)
```

```python
sorted_fruits = sorted(fruits)
print(sorted_fruits)
```

```python
joseon_kings = [
    "태조", "정종", "태종", "세종", "문종", "단종", "세조", "예종", "성종",
    "연산군", "중종", "인종", "명종", "선조", "광해군", "인조", "효종", "현종", "숙종",
    "경종", "영조", "정조", "순조", "헌종", "철종", "고종", "순종"
]
```

```python
for king in joseon_kings:
  print(king)
```

```python
for i, king in enumerate(joseon_kings):
  print(f"{i+1}대왕, {king}")
```

```python
name = (input('왕을 입력하시오 : '))
if name in joseon_kings:
  k = joseon_kings.index(name)
  print(f"{k+1}대왕, {name}")
else:
    print("목록에 없는 왕 이름입니다.")
```

```python
for i in range (len(joseon_kings)-1,-1,-1):
  print(joseon_kings[i])
```
