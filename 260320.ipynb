```python
# 사용자에게 정수를 입력받는다
number = int(input("숫자를 입력하세요: "))

# 2로 나눈 나머지가 0이면 짝수, 1이면 홀수
if number > 0:
    print("양수입니다.")
else:
    print("음수입니다.")
```

```python
coffee_price=2000
latte_price=3000
matcha_price=2500
juice_price=2800

coffee=int(input('커피 주문 수 :'))
latte=int(input('라떼 주문 수 :'))
matcha=int(input('녹차 주문 수 :'))
juice=int(input('주스 주문 수 :'))

price=(coffee_price*coffee+latte_price*latte+matcha_price*matcha+juice_price*juice)
print(f'주문금액은 {price} 입니다.')
```
